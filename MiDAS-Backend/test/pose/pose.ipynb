```python
import json
from pathlib import Path
import requests
import base64
from io import BytesIO
from PIL import Image

# -----------------------
# CONFIG
# -----------------------
SERVER_URL = "http://localhost:8000/pose"
IMAGE_PATH = "pose_01.jpg"          # path to a local image
JSON_PATH = "pose_01.json"

def load_meta(meta_path: Path) -> dict:
    return json.loads(meta_path.read_text(encoding="utf-8"))

# -----------------------
# SEND REQUEST
# -----------------------
with open(IMAGE_PATH, "rb") as f:
    files = {
        "image": ("test.jpg", f, "image/jpeg"),
    }
    data = {
        "payload": json.dumps(load_meta(Path(JSON_PATH)))
    }

    print("[CLIENT] Sending request...")
    resp = requests.post(SERVER_URL, files=files, data=data, timeout=120)

if resp.status_code != 200:
    print("[CLIENT] Request failed!")
    print(resp.status_code, resp.text)
    raise SystemExit(1)

result = resp.json()
print("[CLIENT] Response keys:", result.keys())
# print("[CLIENT] Num instances:", result["returned"])
```

```python
image_bytes = base64.b64decode(result['img_png'])
image = Image.open(BytesIO(image_bytes))
image.show()
```
