```python
from pathlib import Path
import requests
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
```

```python
SERVER_URL = "http://localhost:8000/infer/glb"
IMAGE_PATH = "images/image.png"
MASK_PATH = "images/mask.png"


# -----------------------
# SEND REQUEST
# -----------------------
with open(IMAGE_PATH, "rb") as image, open(MASK_PATH, "rb") as mask:
    files = {
        "image": ("image.png", image, "image/png"),
        "mask": ("mask.png", mask, "image/png")
    }

    print("[CLIENT] Sending request...")
    resp = requests.post(SERVER_URL, files=files, timeout=120)

if resp.status_code != 200:
    print("[CLIENT] Request failed!")
    print(resp.status_code, resp.text)
    raise SystemExit(1)
```

```python
if resp.status_code == 200:
    # Save the received GLB bytes to a file locally
    with open("output_result.glb", "wb") as f:
        f.write(resp.content)
    print("GLB file saved as 'output_result.glb'.")
else:
    print(f"Failed to get GLB. Status code: {resp.status_code}")
    print(resp.text)
```
